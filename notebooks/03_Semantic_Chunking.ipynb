```python
import sys
from pathlib import Path

project_root = Path.cwd().parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
```

```python
from src.config import DOCS_PATH
from src.loaders.document_loader import DocumentLoader
```

```python
loader = DocumentLoader(DOCS_PATH)

documents = loader.load()

print(f"Source Documents: {len(documents):,}")
```

```python
from langchain_huggingface import (
    HuggingFaceEmbeddings
)
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-m3"
)
```

```python
from langchain_experimental.text_splitter import (
    SemanticChunker
)
```

```python
semantic_splitter = SemanticChunker(
    embedding_model
)
```

```python
semantic_chunks = semantic_splitter.split_documents(
    documents
)

print(
    f"Total Semantic Chunks: {len(semantic_chunks):,}"
)
```

```python
semantic_lengths = [
    len(chunk.page_content)
    for chunk in semantic_chunks
]

print("=" * 80)

print("SEMANTIC CHUNKING STATISTICS")

print("=" * 80)

print(f"Chunk Count     : {len(semantic_lengths):,}")

print(f"Min Length      : {min(semantic_lengths):,}")

print(f"Max Length      : {max(semantic_lengths):,}")

print(
    f"Average Length  : {sum(semantic_lengths)/len(semantic_lengths):.0f}"
)
```

```python
for i, chunk in enumerate(semantic_chunks[:3]):

    print("\n" + "=" * 80)

    print(f"SEMANTIC CHUNK #{i+1}")

    print("=" * 80)

    print(chunk.page_content[:1500])
```
